```python
#Random Walk Simulator
import random
import matplotlib.pyplot as plt

# Function to simulate symmetric random walk with boundaries, starting at 0.
def random_walk(boundary):
    position = 0
    steps = 0
    # Walk until boundary is hit
    while -boundary < position < boundary:
        step = random.choice([-1, 1])
        position += step
        steps += 1

    return steps, position

# Run multiple simulations and collect results
results = []
hits_positive = 0
hits_negative = 0

for i in range(1000):
    steps, final_pos = random_walk(10)
    results.append(steps)
    
    #Count which boundary was hit
    if final_pos == 10:
        hits_positive += 1
    else:
        hits_negative += 1

# Calculate & print hitting probabilities
prob_positive = hits_positive / 1000
prob_negative = hits_negative / 1000
print("Probability of hitting +10:", prob_positive)
print("Probability of hitting -10:", prob_negative)

# Plot distribution of hitting times
plt.hist(results, bins = 50)
plt.xlabel("Hitting Time")
plt.ylabel("Frequency")
plt.title("Random Walk Hitting Time Distribution")
plt.show

# Random Walk - This section builds a basic Monte Carlo model using a symmetric random walk with absorbing boundaries at ±10.
#The aim is to estimate hitting probabilities and visualise the distribution of hitting times.
#The histogram shows a right-skewed distribution, which is typical for random walks with absorbing states.
#The empirical hitting probabilities are close to 0.5, matching the theoretical expectation for a symmetric walk starting at 0.
```
